### **1.2. Alcance**

**¿Qué solución de procesamiento de lenguaje natural tratamos de implementar?**
Se implementará un *pipeline* analítico dual de Procesamiento de Lenguaje Natural (NLP) orientado al análisis profundo de reseñas de libros. Para extraer el máximo valor del texto no estructurado, la solución integrará dos componentes principales:

* **Análisis de Sentimientos (Supervisado):** Clasificación binaria de la polaridad de las reseñas (Positivo / Negativo), abordando técnicamente el reto del desbalanceo de clases para priorizar la detección de la clase minoritaria.
* **Modelado de Tópicos (No Supervisado):** Descubrimiento de la estructura temática latente en el corpus para identificar automáticamente nichos literarios, aspectos del producto y hábitos de lectura, sin depender de etiquetas predefinidas.

El proyecto desarrollará un flujo de trabajo integral que contempla las siguientes fases:

* **Fase Descriptiva (Tópicos):** Aplicación de algoritmos de factorización matricial continua (NMF sobre matrices TF-IDF) y arquitecturas modulares avanzadas (BERTopic) para segmentar y extraer las palabras clave representativas de cada clúster temático.

Con respecto a documentos con problemas de codificación, sí hay presencia en algunos documentos (0.55%). Esto se sabe buscando caracteres inválidos o mal codificados, como serían: �,Ã, Â, â€™, entre otros, utilizando expresiones regulares en python, como en el siguiente código:

```
import re

def detect_encoding_issues(text):
    if pd.isnull(text):
        return False

    text = str(text)

    replacement_char = "�" in text
    suspicious_sequences = bool(re.search(r"Ã.|Â.|â.", text))
    non_printable = bool(re.search(r"[^\x20-\x7E\n\r\t]", text))

    return replacement_char or suspicious_sequences or non_printable

df["encoding_issue"] = df["review_body"].apply(detect_encoding_issues)

df[df["encoding_issue"]].head()

total_issues = df["encoding_issue"].sum()
percentage_issues = (total_issues / len(df)) * 100

print("Textos con posibles problemas:", total_issues)
print("Porcentaje:", round(percentage_issues, 2), "%")
```

Adicional exploramos acontinuación algunos registros donde evidenciamos caracteres relacionados a html como "\<br\>" , enlaces y expresiones de distinta indole cómo  \\"contemporaries\\"

Es por eso que mediante la siguiente celda evaluamos porcentaje de registros con este tipo de expresiones

```
# ========= PATRONES (detección) =========
PATTERNS = {
    "html_tag": r"<[^>]+>",  # <br>, <a ...>, etc.
    "html_script_style": r"<(script|style)\b[^>]*>",  # <script ...> o <style ...>
    "html_href": r"\bhref\s*=\s*['\"]?[^'\"\s>]+",  # href=...
    "html_entity": r"&[a-zA-Z]+;|&#\d+;|&#x[0-9a-fA-F]+;",  # &amp; &#123; &#x1F;
    "url": r"(https?:[ ]//\S+|www\.\S+)",
    "backslash_any": r"\\",            # cualquier \
    "escaped_newline_tab": r"\\[nrt]", # \n \r \t como texto
    "unicode_escape": r"\\u[0-9a-fA-F]{4}",  # \u2019 etc
}

MAX_EXAMPLES = 5

compiled = {k: re.compile(v, re.IGNORECASE | re.DOTALL) for k, v in PATTERNS.items()}

def analyze_first_n(text, nrows: int = 100_000):
    text = pd.Series(text) # Convert list to Series

    total = len(text)
    rows = []

    examples = {k: [] for k in PATTERNS.keys()}

    for name, pattern in PATTERNS.items():
        mask = text.str.contains(pattern, regex=True, na=False)
        count = int(mask.sum())
        pct = (count / total * 100) if total else 0.0

        rows.append({"pattern": name, "count_rows": count, "pct_rows": pct})

        # Ejemplos (snippets)
        if count > 0:
            sample_series = text[mask].head(MAX_EXAMPLES)
            for s in sample_series.tolist():
                s_short = s.replace("\n", " ")[:300]
                m = compiled[name].search(s)
                hit = (m.group(0)[:80] if m else "")
                examples[name].append({"match": hit, "text_snippet": s_short})

    summary_df = pd.DataFrame(rows).sort_values("count_rows", ascending=False)

    return total, summary_df, examples

def analize_patterns(text):
    total, summary_df, examples = analyze_first_n(text)

    print(f"\nFilas analizadas: {total:,}\n")
    print(summary_df.to_string(index=False, formatters={"pct_rows": "{:.2f}%".format}))

    print("\n\n=== Ejemplos por patrón (primeros) ===")
    for name, exs in examples.items():
        if not exs:
            continue
        print(f"\n[{name}]")
        for i, ex in enumerate(exs, 1):
            print(f"  {i}) match: {ex['match']}")
            print(f"     text : {ex['text_snippet']}")
```

```
import spacy
from spacy.language import Language
from spacy_langdetect import LanguageDetector

#Configurar modelo de lenguaje
spacy.cli.download("en_core_web_sm")
nlp_detector = spacy.load("en_core_web_sm")

#Configurar detector de idioma
def get_lang_detector(nlp, name):
    return LanguageDetector()
try:
  Language.factory("language_detector", func=get_lang_detector)
except:
  pass

nlp_detector.add_pipe('language_detector', last=True)
```

```
from tqdm import tqdm

def detectar_idioma_batch(text_list):
    idiomas = []

    # Asegúrate de convertir a string por si hay algún nulo colado
    textos_str = [str(t) for t in text_list]

    # Usamos nlp.pipe para procesar en lotes reales
    generador = nlp_detector.pipe(textos_str, batch_size=1000)

    # tqdm nos da la barra de progreso
    for doc in tqdm(generador, total=len(textos_str), desc="Detectando idioma"):
        # Extraemos el idioma del documento
        idiomas.append(doc._.language['language'])

    return idiomas

# Lo aplicas creando una nueva columna en tu DataFrame
df['idioma'] = detectar_idioma_batch(df['review_body'].tolist())
```

A partir de este texto depurado, nuestra estrategia se bifurcará en dos flujos de procesamiento (*pipelines*) adaptados a las necesidades específicas de cada técnica de NLP:

* **Flujo A (Contextual - Transformers):** Orientado al análisis de sentimiento y emociones. En este caso, aplicaremos una limpieza mínima para preservar la puntuación, las negaciones y la estructura gramatical. Esto es vital para que los modelos profundos puedan interpretar correctamente el tono, el contexto y la intensidad emocional del lector.
* **Flujo B (Agresivo - Modelos Clásicos):** Orientado a la extracción de temas (*Topic Modeling*) . Aquí aplicaremos un procesamiento estricto que incluye normalización ASCII, lematización y eliminación de *stopwords*. El objetivo es reducir la sintaxis para quedarnos únicamente con la raíz de los conceptos más relevantes, disminuyendo la dimensionalidad del vocabulario y optimizando la creación de matrices para algoritmos como TF-IDF, LDA o NMF.

```
import re
import html

# ========= PATRONES (Pre-compilados para velocidad) =========
RE_SCRIPT_STYLE = re.compile(r"<(script|style)\b[^>]*>.*?</\1>", re.IGNORECASE | re.DOTALL)
RE_TAGS         = re.compile(r"<[^>]+>")
RE_AMAZON_TAG   = re.compile(r"\[\[ASIN:.*?\]\]", re.IGNORECASE)
RE_URL          = re.compile(r"(https?://\S+|www\.\S+)")
RE_MULTI_DOT = re.compile(r"\.{2,}")

RE_BS_QUOTE     = re.compile(r"\\([\"'])")
RE_BS_OTHER     = re.compile(r"\\+")

RE_FANCY_QUOTES = re.compile(r"[“”]")
RE_FANCY_APOS   = re.compile(r"[’‘]")

# Permite letras con tildes, números, espacios y puntuación esencial
RE_KEEP_EN      = re.compile(r"[^A-Za-zÀ-ÿ0-9\s\.\,\!\?\;\:\'\"\-\(\)\[\]]+")
RE_MULTI_SPACE  = re.compile(r"\s{2,}")

def simple_preprocess(text):

    # 1. Decodificar entidades HTML (ej. &amp; -> &)
    text = html.unescape(text)

    # 2. Reemplazar tags HTML y saltos de línea por espacios
    text = RE_SCRIPT_STYLE.sub(" ", text)
    text = RE_TAGS.sub(" ", text)
    text = text.replace("\\n", " ").replace("\\r", " ").replace("\\t", " ")

    # 3. Eliminar tags de Amazon (ASIN)
    text = RE_AMAZON_TAG.sub(" ", text)

    # 4. Eliminar URLs
    text = RE_URL.sub(" ", text)

    # 5. Normalizar comillas tipográficas a estándar
    text = RE_FANCY_QUOTES.sub('"', text)
    text = RE_FANCY_APOS.sub("'", text)

    #Comprimir múltiples puntos en un solo punto con un espacio
    text = RE_MULTI_DOT.sub(". ", text)

    # 6. Des-escapar comillas y eliminar backslashes huérfanos
    text = RE_BS_QUOTE.sub(r"\1", text)
    text = RE_BS_OTHER.sub(" ", text)

    # 7. Mantener solo caracteres permitidos (filtro final)
    text = RE_KEEP_EN.sub(" ", text)

    # 8. Colapsar espacios múltiples y limpiar extremos
    text = RE_MULTI_SPACE.sub(" ", text).strip()

    return text
```

```
import spacy
from unidecode import unidecode
from tqdm import tqdm

nlp = spacy.load("en_core_web_sm", disable=["parser", "ner"])

def batch_preprocess(text_list):
    clean_texts = []

    generador_spacy = nlp.pipe(text_list, batch_size=1000, n_process=-1)

    for doc in tqdm(generador_spacy, total=len(text_list), desc="Procesando reseñas"):
        tokens_utiles = []
        for token in doc:
            # Filtro base
            if not token.is_punct and not token.is_stop and token.is_alpha:
                lema = token.lemma_.lower()

                # Filtro por longitud
                if 2 <= len(lema) <= 20:
                    tokens_utiles.append(lema)

        # Unimos y normalizamos
        texto_unido = " ".join(tokens_utiles)
        texto_normalizado = unidecode(texto_unido)

        clean_texts.append(texto_normalizado)

    return clean_texts
```

## **1. Selección del Embedding**
---

Seleccione el tipo de **embedding** que considere más apropiado para su problema. Recuerde las siguientes consideraciones:

- **Bolsas de palabras**: son útiles para representar documentos completos. Funcionan bien con corpus pequeños (<10000 documentos).
- **TF-IDF**: permite obtener una representación ponderada de documentos completos; es una alternativa a las bolsas de palabras.
- **Bolsas de N-grams**: a nivel de carácter son útiles para representar palabras o segmentos cortos de texto, funcionan bien con corpus pequeños (<10000 documentos).
- **Word2Vec**: permite representar palabras de forma semántica, requiere corpus grandes para un buen resultado (>10000 documentos).
- **FastText**: permite representar palabras o segmentos cortos de texto, funcionan bien con corpus grandes (>10000 documentos). Es una alternativa a word2vec que puede representar palabras fuera del vocabulario.
- **Doc2Vec**: permite representar documentos completos de forma semántica. Funciona bien con corpus grandes (>10000 documentos).


Para abordar de manera integral los objetivos del proyecto (extracción de temas y análisis de sentimiento) sobre un corpus masivo superior a los 100.000 documentos, se ha seleccionado un enfoque dual que aprovecha los dos flujos de preprocesamiento previamente establecidos:

**1. TF-IDF (Term Frequency - Inverse Document Frequency)**
* **Asignado a:** Flujo de limpieza agresiva (texto lematizado, sin *stopwords*).
* **Caso de uso:** Extracción de Temas (*Topic Modeling*)
* **Justificación:** Al contar con un corpus grande (>10.000 documentos), una simple Bolsa de Palabras (BoW) o N-gramas generarían matrices excesivamente densas y ruidosas donde las palabras más comunes opacarían a las verdaderamente relevantes. Se selecciona **TF-IDF** porque permite obtener una representación ponderada a nivel de documento completo, penalizando términos genéricos y otorgando mayor peso matemático a las palabras específicas que definen la temática de cada reseña



**2. Doc2Vec / Embeddings Contextuales**
* **Asignado a:** Flujo de limpieza contextual (texto simple con puntuación y gramática intacta).Clasificación.
* **Justificación:** El sentimiento de un lector no recae en palabras aisladas, sino en la semántica global de la reseña. Por ello, enfoques a nivel de palabra como Word2Vec resultan insuficientes, ya que promediar sus vectores diluye el contexto general del texto. Se selecciona **Doc2Vec** (o los embeddings propios de un modelo Transformer) porque permite representar documentos completos de forma semántica. Al tener un corpus grande (>10.000 documentos), el modelo cuenta con datos suficientes para mapear reseñas enteras en un espacio vectorial, capturando la polaridad y el tono del texto de manera mucho más precisa que las representaciones basadas en frecuencias.

## **2.1 Partición de Datos (Train, Validation, Test)**
---
Antes de proceder con la vectorización y extracción de características, es un principio fundamental del Machine Learning realizar la partición del dataset.

**¿Por qué particionar antes de vectorizar?**
Si aplicáramos TF-IDF o Doc2Vec a todo el dataset completo y luego hiciéramos la división, cometeríamos **Fuga de Datos (*Data Leakage*)**. El modelo aprendería el vocabulario y la distribución de palabras del set de prueba, arruinando la validez de nuestra evaluación futura.



Dado el fuerte desbalanceo de clases detectado en la variable objetivo (85% Positivo, 9% Negativo, 5% Neutral), utilizaremos el parámetro `stratify` para garantizar que esta proporción se mantenga exacta en los tres conjuntos:
* **Entrenamiento (Train):** 70% (Para enseñar a los modelos).
* **Validación (Val):** 15% (Para ajustar hiperparámetros y tomar decisiones de balanceo).
* **Prueba (Test):** 15% (Conjunto completamente ciego para la evaluación final).


Adicional para el caso del flujo de categorización, por simplicidad vamos a quedarnos únicamente con las reviews positivas y negativas, vamos a filtrar las neutrales.

## **2.3 Extracción de Características: Implementación de Doc2Vec**
---
Este pipeline procesará la columna `cleaned_review_simple` (texto contextualizado) para generar embeddings densos que representen semánticamente cada reseña completa, ideal para el Análisis de Sentimientos.



Utilizaremos la librería `gensim`. El proceso consta de dos fases:
1.  **Entrenamiento del Modelo:** El algoritmo Doc2Vec construirá su propio vocabulario y aprenderá a vectorizar documentos usando exclusivamente el set de entrenamiento.
2.  **Inferencia:** Utilizaremos el modelo entrenado para "inferir" los vectores matemáticos de los textos de Validación y Prueba.

```
%pip install gensim
```

```
from gensim.models.doc2vec import Doc2Vec, TaggedDocument
from tqdm import tqdm

# 1. Preparar los datos de Train en el formato TaggedDocument
train_tagged = [
    TaggedDocument(words=str(texto).split(), tags=[str(i)])
    for i, texto in enumerate(X_train_bin['cleaned_review_simple'])
]

# 2. Inicializar y entrenar el modelo Doc2Vec (vector_size=100 dimensiones)
print("Construyendo vocabulario y entrenando modelo Doc2Vec...")
doc2vec_model = Doc2Vec(vector_size=200, window=10, min_count=5, workers=-1, epochs=200)
doc2vec_model.build_vocab(train_tagged)
doc2vec_model.train(train_tagged, total_examples=doc2vec_model.corpus_count, epochs=doc2vec_model.epochs)

# 3. Función para inferir vectores (transformar texto en números usando el modelo)
def generar_vectores_d2v(modelo, text_series, desc_text="Infiriendo vectores"):
    vectores = []
    for texto in tqdm(text_series, desc=desc_text):
        vector = modelo.infer_vector(str(texto).split())
        vectores.append(vector)
    return vectores

# 4. Generamos las matrices finales para los 3 conjuntos
X_train_d2v = generar_vectores_d2v(doc2vec_model, X_train_bin['cleaned_review_simple'], "Infiriendo Train")
X_val_d2v   = generar_vectores_d2v(doc2vec_model, X_val_bin['cleaned_review_simple'], "Infiriendo Val")
X_test_d2v  = generar_vectores_d2v(doc2vec_model, X_test_bin['cleaned_review_simple'], "Infiriendo Test")
```

```
# Guardar los vectores Doc2Vec como CSV
print("Guardando vectores Doc2Vec en CSV...")
pd.DataFrame(X_train_d2v).to_csv('X_train_d2v.csv', index=False)
pd.DataFrame(X_val_d2v).to_csv('X_val_d2v.csv', index=False)
pd.DataFrame(X_test_d2v).to_csv('X_test_d2v.csv', index=False)
print("Vectores Doc2Vec guardados exitosamente.")
```

Para superar las limitaciones espaciales y semánticas de los enfoques clásicos, se propone la implementación de un modelo de estado del arte (SOTA) basado en arquitecturas de *Deep Learning*.

A diferencia de las representaciones estáticas como *Doc2Vec*, DistilBERT procesa el texto en su formato original utilizando un **mecanismo de atención bidireccional**.  Esto le permite evaluar el contexto completo de una oración de manera simultánea, siendo altamente efectivo para capturar matices complejos del lenguaje, como negaciones invertidas, sarcasmos y dependencias a largo plazo.

### **3.2.1 Tipo de Modelamiento y Selección**
El objetivo de esta fase es descubrir la estructura temática latente dentro del corpus de reseñas, sin utilizar etiquetas predefinidas.

* **Algoritmo Seleccionado:** `Non-negative Matrix Factorization (NMF)`.
* **Características:** Matriz dispersa generada por **TF-IDF** sobre el texto avanzado (lematizado y sin *stopwords*).
* **Justificación:** Aunque LDA es popular, es un modelo probabilístico que asume conteos enteros. Al estar utilizando TF-IDF (que genera valores decimales continuos), la descomposición de matrices de NMF es matemáticamente la opción correcta. Además, al restringir los valores a números no negativos, NMF genera tópicos estrictamente aditivos e interpretables .

```
# --- IMPORTACIONES PARA TÓPICOS ---
from sklearn.decomposition import NMF
import matplotlib.pyplot as plt



def get_NMF_model(X_train_tfidf, n_components):
    nmf_model = NMF(
        n_components=n_components,
        random_state=42,
        init='nndsvda',
        max_iter=500)

    print("="*60)
    print("ENTRENAMIENTO DE MODELO DE TÓPICOS (NMF)")
    print("="*60)
    nmf_model.fit(X_train_tfidf)
    print("¡Modelo NMF entrenado con éxito!")
    return nmf_model


n_components = 3


nmf_model = get_NMF_model(X_train_tfidf, n_components)
```

```
# Definimos la función de graficado optimizada para el ancho del notebook
def plot_top_words(model, feature_names, n_top_words, title):
    # Ajustamos el figsize para que se vea perfecto en la pantalla
    fig, axes = plt.subplots(1, n_components, figsize=(25, 10), sharex=True)
    axes = axes.flatten()

    for topic_idx, topic in enumerate(model.components_):
        # Extraemos los índices de las palabras con más peso
        top_features_ind = topic.argsort()[-n_top_words:]
        top_features = feature_names[top_features_ind]
        weights = topic[top_features_ind]

        # Graficamos las barras horizontales
        ax = axes[topic_idx]
        ax.barh(top_features, weights, height=0.7, color='steelblue')
        ax.set_title(f"Tópico {topic_idx + 1}", fontdict={"fontsize": 22, "fontweight": "bold"})
        ax.tick_params(axis="both", which="major", labelsize=14)

        # Limpiamos los bordes para un diseño más moderno
        for i in "top right left".split():
            ax.spines[i].set_visible(False)

    fig.suptitle(title, fontsize=30, fontweight="bold")
    plt.subplots_adjust(top=0.85, bottom=0.05, wspace=0.5, hspace=0.3)
    plt.show()

# 1. Extraemos el vocabulario de nuestro vectorizador TF-IDF original
vocabulario_tfidf = tfidf.get_feature_names_out()

# 2. Generamos la gráfica
plot_top_words(
    model=nmf_model,
    feature_names=vocabulario_tfidf,
    n_top_words=15,
    title="Palabras Clave por Tópico (Modelo NMF)"
)
```

**Modelado de Tópicos Avanzado (Estado del Arte): BERTopic**

Con el propósito de superar las limitaciones inherentes a la vectorización estática (TF-IDF) y capturar la semántica profunda de los textos, se propone la implementación de **BERTopic**, un algoritmo de modelado de tópicos de estado del arte (SOTA).

A diferencia de los enfoques matemáticos clásicos como NMF o LDA, que dependen de la frecuencia absoluta de las palabras, BERTopic es capaz de agrupar documentos basándose en su significado contextual. Para lograr esto, el algoritmo no opera en un solo paso matemático, sino que emplea un *pipeline* modular compuesto por cuatro fases secuenciales:

1. **Generación de *Embeddings* Contextuales:** Utiliza modelos de lenguaje pre-entrenados basados en la arquitectura *Transformer* (como *Sentence-Transformers*) para convertir cada documento en un vector denso que retiene el contexto bidireccional de las oraciones.
2. **Reducción de Dimensionalidad (UMAP):** Dado que los *embeddings* operan en espacios de alta dimensión, se aplica el algoritmo UMAP (*Uniform Manifold Approximation and Projection*) para reducir la dimensionalidad del espacio vectorial, preservando simultáneamente la estructura semántica local y global de los datos.
3. **Agrupamiento basado en Densidad (HDBSCAN):** A diferencia de NMF, donde el investigador debe definir arbitrariamente el número de clústeres ($k$), BERTopic emplea HDBSCAN (*Hierarchical Density-Based Spatial Clustering of Applications with Noise*). Este algoritmo descubre automáticamente la cantidad óptima de tópicos basándose en la densidad espacial de los vectores y aísla las reseñas anómalas o genéricas en un clúster de ruido (Tópico `-1`), evitando que contaminen los temas principales.
4. **Extracción de Representación (c-TF-IDF):** Finalmente, para dotar de interpretabilidad humana a cada clúster, el modelo aplica una variante basada en clases del TF-IDF tradicional (*class-based TF-IDF*). Esto permite extraer las palabras clave más representativas de cada tópico descubierto.


```
print("="*60)
print("BERTopic ACELERADO POR GPU")
print("="*60)


documentos_lista = [str(texto) for texto in X_train['cleaned_review_simple'].tolist()]

# 1. Inicializamos el modelo de lenguaje explícitamente en la GPU
print("Cargando modelo Transformer en la tarjeta gráfica...")
# "all-MiniLM-L6-v2" es el modelo SOTA por defecto de BERTopic:
modelo_embeddings = SentenceTransformer("all-MiniLM-L6-v2", device="cuda") # o device=0

# 2. Le pasamos nuestro modelo acelerado a BERTopic
topic_model = BERTopic(
    embedding_model=modelo_embeddings,
    min_topic_size=50,
    verbose=True
)

# 3. Entrenar el modelo
print("Extrayendo embeddings y calculando tópicos (Esto será rápido)...")
topics, probabilities = topic_model.fit_transform(documentos_lista)

print("\n¡Modelo BERTopic entrenado a máxima velocidad!")

df_info_topicos = topic_model.get_topic_info()
display(df_info_topicos.head(10))
```

El presente proyecto abordó el análisis de un corpus masivo de reseñas literarias mediante un enfoque dual de *Procesamiento de Lenguaje Natural* (NLP): la clasificación supervisada de sentimientos y el descubrimiento no supervisado de tópicos latentes. La evaluación empírica de múltiples algoritmos arroja las siguientes conclusiones fundamentales:

**1. El Cambio de Paradigma en NLP: Vectorización Estática vs. Contexto Dinámico**
A lo largo de la fase de clasificación, se demostró que las técnicas de representación vectorial clásica (como *TF-IDF* y *Doc2Vec*) presentan un techo de rendimiento estricto al enfrentarse a semántica compleja (sarcasmo, negaciones invertidas). Los modelos tradicionales, como la Regresión Logística y Random Forest, no lograron trazar fronteras de decisión eficientes en este espacio geométrico. Por el contrario, la transición hacia arquitecturas de *Deep Learning* (DistilBERT) demostró que el uso de mecanismos de atención bidireccional es indispensable para capturar el significado real de las oraciones en su contexto nativo, cuadruplicando la capacidad predictiva sobre las clases más difíciles.

**2. Resolución de la Paradoja de la Exactitud en Datos Desbalanceados**
El análisis evidenció los peligros de optimizar métricas globales (como el *Accuracy*) en datasets altamente desbalanceados (90% positivos / 10% negativos). Mientras que los modelos de ensamble clásicos ignoraron la clase minoritaria para inflar su exactitud global, el modelo *Transformer* logró un hito analítico: alcanzar un *Recall* del 89% en la detección de quejas (clase negativa) manteniendo un 90% de exactitud general, todo ello sin requerir técnicas artificiales de balanceo de pesos. Esto viabiliza la detección automatizada de áreas de mejora para el negocio.

**3. La Evolución del Modelado de Tópicos hacia la Autonomía**
En la fase de aprendizaje no supervisado, el análisis iterativo con el algoritmo NMF validó empíricamente que $k=7$ es el número óptimo de componentes para extraer macrogéneros literarios y patrones de consumo. Sin embargo, la implementación de la arquitectura modular **BERTopic** revolucionó la granularidad del análisis. Al prescindir de agrupamientos forzados e integrar reducción de dimensionalidad (UMAP) y clustering espacial (HDBSCAN), el modelo SOTA logró descubrir de forma 100% autónoma micro-nichos de mercado altamente específicos (ej. libros de cocina, religión, novela negra) aislando eficientemente el ruido semántico.

**4. El Impacto Crítico del Preprocesamiento Específico de Dominio**
Tanto en la fase supervisada como en la no supervisada, se evidenció que la limpieza estándar de texto es insuficiente. La inspección visual de los clústeres reveló que la retención de *stopwords* de dominio (términos genéricos como *"book"* o *"read"*) asfixia la varianza estadística del corpus y oculta palabras clave secundarias de alto valor. La exclusión proactiva de este vocabulario redundante es un paso crítico e innegociable para garantizar la pureza temática de los modelos.
